###Accediendo a Azure Data Lake Storage mediante Access Key

####Establecemos la configuracion necesaria

```python
access_key = dbutils.secrets.get(scope = "movie-secrets-scope", key = "movie-access-key")
```

```python
spark.conf.set("fs.azure.account.key.projectmovies.dfs.core.windows.net", access_key)
```

####Listamos archivos

```python
display(dbutils.fs.ls("abfss://demo@projectmovies.dfs.core.windows.net"))
```

####Leemos el archivo

```python
spark.read.option("header", "true").csv("abfss://demo@projectmovies.dfs.core.windows.net/movie.csv")
```
