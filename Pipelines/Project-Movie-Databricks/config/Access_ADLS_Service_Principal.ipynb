###Accediendo a Azure Data Lake Storage mediante Service Principal

####Establecemos la configuracion necesaria

```python
#Sacamos estos datos registrando una app para nuestro proyecto en Micrososft Entra ID
#Y accediendo a ella mediante Key Vault y Secrets Scope
client_id = dbutils.secrets.get(scope = "movie-secrets-scope", key = "movie-service-principal-client-id")
tenant_id = dbutils.secrets.get(scope = "movie-secrets-scope", key = "movie-service-principal-tenant-id")
#En la app registrada creamos una client secret
client_secret = dbutils.secrets.get(scope = "movie-secrets-scope",key = "movie-service-principal-client-secret")
```

```python
spark.conf.set("fs.azure.account.auth.type.projectmovies.dfs.core.windows.net", "OAuth")
spark.conf.set("fs.azure.account.oauth.provider.type.projectmovies.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider")
spark.conf.set("fs.azure.account.oauth2.client.id.projectmovies.dfs.core.windows.net", client_id)
spark.conf.set("fs.azure.account.oauth2.client.secret.projectmovies.dfs.core.windows.net", client_secret)
spark.conf.set("fs.azure.account.oauth2.client.endpoint.projectmovies.dfs.core.windows.net", f"https://login.microsoftonline.com/{tenant_id}/oauth2/token")
```
