###Accediendo a Azure Data Lake Storage mediante Token SAS

####Establecemos la configuracion necesaria

```python
#Traemos el token sas generado para el contenedor demo que esta cifrado por key vault y secrets scope
token_sas = dbutils.secrets.get(scope = "movie-secrets-scope", key = "movie-token-sas")
```

```python
spark.conf.set("fs.azure.account.auth.type.projectmovies.dfs.core.windows.net", "SAS")
spark.conf.set("fs.azure.sas.token.provider.type.projectmovies.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.sas.FixedSASTokenProvider")
spark.conf.set("fs.azure.sas.fixed.token.projectmovies.dfs.core.windows.net", token_sas)
```
